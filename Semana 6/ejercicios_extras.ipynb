```python
#ejercicio 10

suma = 0

while True:
    numero = int(input("Ingresa un numero (0 para terminar): "))
    if numero == 0:
        break
    suma += numero

print(f"La suma de los números es: {suma}")
```

```python
#ejercicio 12

suma = 0

for i in range(1, 21):
    if i % 2 != 0:
        continue
    suma += i

print(f"La suma de los números pares del 1 al 20 es: {suma}")
```

```python
#ejercicio 13

def sumar_lista(lista):
    total = 0
    for elemento in lista:
        total += elemento
    return total

elemento = [10, 20, 30, 40]
resultado = sumar_lista(elemento)
print(f"La suma de los elementos es: {resultado}")
 
```

```python
#ejercicio 14

def obtener_maximo(lista):
    maximo = lista[0]
    for num in lista:
        if num > maximo:
            maximo = num
    return maximo

elementos = [7, 3, 9, 2]
resultado = obtener_maximo(elementos)
print(f"El elemento máximo es: {resultado}")
```

```python
#ejercicio 15

matriz =  [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
           ]

suma_diagonal = 0
for i in range(len(matriz)):
    suma_diagonal += matriz[i][i]

print(f"La suma de la diagonal principal es: {suma_diagonal}")
```

```python
#ejercicio 16

def imprimi_promedio(calificacion):
    suma = 0
    for cal in calificacion:
        suma += cal
    promedio = suma / len(calificacion)
    print(f"El promedio es: {promedio:.2f}")

imprimi_promedio([8, 9, 10, 7])
```

```python
#ejercicio 18

def factorial(n):
    if n == 0 or n == 1:
        return  1
    return n * factorial(n - 1)

numero = 5
resultado = factorial(numero)
print(f"El factorial de {numero} es: {resultado}")
```

```python
#ejercicio 19 

def sumar_tupla(tup):
    return sum(tup)

tupla_datos = (4, 8, 15, 23, 42)

print(f"El tercer elemento es: {tupla_datos[2]}")
print(f"La suma de la tupla es: {sumar_tupla(tupla_datos)}")
```

```python
#ejercicio 21

estudiantes = {
    "Ana": [8, 9, 10],
    "Beto": [7, 7, 8]
}

for nombre, notas in estudiantes.items():
    promedio = sum(notas) / len(notas)
    print(f"{nombre}: {promedio:.2f}")
```
