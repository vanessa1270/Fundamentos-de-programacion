```python
#ejercicio 2

candidatos = {
    "A": 0,
    "B": 0,
    "C": 0
}

cantidad = int(input("Ingresa la cantidad de votos que serán registrados: "))

conteo = 1

while conteo <= cantidad:
    voto = input(f"Ingresa tu voto {conteo} (A, B, C): ")
    
    if voto.upper() in candidatos:
        print(f"Voto válido para {voto.upper()}")
        candidatos[voto.upper()] += 1
        conteo += 1
    else:
        print("El voto no es válido")


print("\nResultados")
for k, v in candidatos.items():
    print(f" Opción {k}, votos {v} ")

```

```python
#ejercicio 5

contactos = {
    "Leo": "9981274595",
    "Gerardo": "9980345623",
    "Santi": "9982737465"
}

numeros_tupla = (8, 4, 1, 9, 2)


def tupla_ordenada():
    print(f"tupla ordenada: {sorted(numeros_tupla)}")

def buscar_contacto():
    nombre = input("Ingresa el nombre del contacto: ")
    if nombre in contactos:
        print(f"El télefono de {nombre} es: {contactos[nombre]}")
    else:
        print("Contacto no encontrado, ingresa un contacto existente")

def dividir_numero():
    try:
        num1 = int(input("Ingresa un número entero: "))|
        num2 = int(input("Ingresa un número entero: "))
        division = num1 / num2
    except ZeroDivisionError:
        print("Error: Debes ingresar un número entero.")
    except ValueError:
        print("Error: No se puede dividir entre cero.")
    else:
        print(f"El resultado de la división {division} ")

def analizar_frase():
        texto = input("Ingresa una frase: ")
        print(f"Longitud de la frase: {len(texto)}, Palabras: {len(texto.split())}")

while True:
    print("\t------Menú principal------")
    print("\t1. Mostrar tupla ordenada")
    print("\t2. Buscar télefono en contactos")
    print("\t3. Dividir dos números ")
    print("\t4. Analizar frase")
    print("\t5. Salir del programa")

    opcion = input("Selecciona una opción (1-5): ")

    if opcion == "1":
        tupla_ordenada()
    elif opcion == "2":
        buscar_contacto()
    elif opcion == "3":
        dividir_numero()
    elif opcion == "4":
        analizar_frase()
    elif opcion == "5":
        print("Programa finalizado")
        break
    else:
        print("Elige una opción válida del 1 al 5")
```

```python
#ejercicio 1 

def sumar_calificaciones(calificaciones):
    return sum(calificaciones)

calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)

print(f"La tercera calificación es:{calificaciones[2]}")

n1 = float(input("Agrega un número: "))
n2 = float(input("Agrega un segundo número: "))
nueva_calificacion = calificaciones + (n1, n2)

print(f"La nueva tupla: {nueva_calificacion}")


lista = list(nueva_calificacion)
lista.sort(reverse=True)

print(f"Lalista de mayor a menor es: {lista}")

suma_final = sumar_calificaciones(nueva_calificacion)

print(f"La suma de todas las calificaciones es: {suma_final}")
```

```python
#ejercicio 2

agenda = {
    "Ana": "555-0101", 
    "Luis": "555-0102",
    "Mía": "555-0103"
    }

nuevo = input("Ingresa el nombre del nuevo contacto: ")
tel = input("Ingresa el télefono del nuevo contacto: ")
nombre = input("Ingresa el nombre del contacto a buscar: ")

agenda[nuevo] = tel
nombres = ", ". join (agenda.keys())
print(f"Contacto registrado: {nombres}")

def buscar_contacto(contactos,nombre):
    return contactos.get(nombre, "No existe el contacto")

resultado = buscar_contacto(agenda, nombre)
print(f"El teléfono de {nombre} es: {resultado}")
```
