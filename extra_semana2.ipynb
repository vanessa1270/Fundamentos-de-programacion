```python
#Contro de aforo con break, continue y while (ejercicio 1)

total = 0
boletos = 1

while True:
    costo = int(input(f"Costo del boleto {boletos}: "))
    boletos += 1

    if costo == 0:
        print("Menor de 3 años, sin cargo")
        continue

    total += costo

    if total >=100:
        print("Aforo alcanzado, se detiene el registro")
        break

print(f"Total acumulado {total:.2f}")


    
```

Para realizar este codigo utilice while ya que asi lo pide la instrucción pero primero analice que si el costo es 0 es porq el niño es menor ya que ellos no pagan pero si es mayor a 18 y alcanza un total mayor a 100 se detiene el programa, de esta manera me diola salida esperada

```python
#Estadística de Visitantes (ejercicio 2)
mayor_edad = 0
suma_edades = 0

visitantes = int(input("Cuantos visitantes son: "))
for i in range (1, visitantes +1):

    edad = int(input(f"Cual es la edad del visitante {i}: "))

    suma_edades += edad 

    if edad >=18:
        mayor_edad += 1

promedio_edades = suma_edades / visitantes
print(f"\nAdultos: {mayor_edad}")
print(f"Promedio de edad: {promedio_edades:.2f}")
```

Para este codigo primero pense en utilizar un for para poder hacer la pregunta de cada edad por visitante, des esta manera ya solo le dije como sacar el promedio de la edad de todos para la salida esperada.
