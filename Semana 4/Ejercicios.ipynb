```python
#Preguntar cuantas calificaciones desea promediar, y promediarlas

def promedio(suma, n):
    return suma/n

n = int(input("¿Cuántas calificaciones deseas promediar? "))

suma = 0
for cal in range(1, n+1):
    calificacion = float(input(f"Ingresa la calificación {cal}: "))
    suma = suma + calificacion

print(f"El promedio de  {n} calificaciones es {promedio(suma, n)}")
```

```python
filas = 3
columnas = 3
matriz = []
for f in range (filas):
    fila_actual = []
    for c in range(columnas):
        valor = int(input(f"Valor para [{f}][{c}] "))
        fila_actual.append(valor)
    matriz.append(fila_actual)
```

```python

dimension = int(input("¿Cuál es la dimensión de la matriz? "))
matriz = []
for f in range (dimension):
    fila_actual = []
    for c in range(dimension):
        valor = 0
        if f == c:
            valor = 1
        fila_actual.append(valor)
    matriz.append(fila_actual)

print("Impresion de matriz")
for f in range (dimension):
    for c in range(dimension):
        print(matriz[f][c], end=" ")
    print() 
```
