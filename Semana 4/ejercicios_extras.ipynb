```python
#ejercicio 1

def suma_matriz(matriz):
    suma = 0
    for fila in matriz:
        for elemento in fila:
            suma += elemento
    return suma

def imprimir_matriz(matriz):
    for fila in matriz:
        for elemento in fila:
            print(elemento, end=" ")
        print()

matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print("Matriz:")
imprimir_matriz(matriz)
resultado = suma_matriz(matriz)
print("Suma de los elementos de la matriz:", resultado)
```

```python
#ejercicio 2

def suma_fila(fila):
    suma = 0
    for elemento in fila:
        suma += elemento
    return suma

def mostrar_tabla(matriz):
    for fila in matriz:
        for elemento in fila:
            print(elemento, end=" ")
        print()

matriz = [[3, 1, 4], [1, 5, 9], [2, 6, 5]]
mostrar_tabla(matriz)

for i in range(len(matriz)):
    suma = suma_fila(matriz[i])
    print("Suma de la fila", i + 1, ":", suma)
```

```python
#ejercicio 3

def multiplicar_columna(matriz, columna):
    producto = 1

    for fila in matriz:
        numero = fila[columna]
    
        resultado = 0

        for i in range(numero):
            resultado = resultado + producto

        producto = resultado
    return producto

matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9], [2, 4, 6]]

columna = int(input("Ingrese el número de columna: "))
resultado = multiplicar_columna(matriz, columna)
print("El producto de los elementos de la columna", columna, "es:", resultado)
```

```python
#ejercicio 4

def generar_tabla_pitagoras(n):
    tabla = []
    for fila in range(1, n + 1):
        renglon = []
        for columna in range(1, n + 1):
            producto = 0
            for i in range(columna):
                producto = producto + fila
            renglon.append(producto)
        tabla.append(renglon)
    return tabla

def imprimir_tabla(tabla):
    for fila in tabla:
        for elemento in fila:
            print(elemento, end="\t")
        print()

tamaño = int(input("Ingrese el tamaño de la tabla de Pitágoras: "))

if tamaño < 2 or tamaño > 5:
    print("El tamaño debe estar entre 2 y 5.")
else:
    tabla_pitagoras = generar_tabla_pitagoras(tamaño)
    imprimir_tabla(tabla_pitagoras)

    renglon = int(input("Ingrese el número de fila: "))
    columna = int(input("Ingrese el número de columna: "))

    resultado = tabla_pitagoras[renglon - 1][columna - 1]

    print("El producto de ", renglon, "x", columna, "es:", resultado)
```
