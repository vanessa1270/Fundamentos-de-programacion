```python
#Calificacion por rango de edad (ejercicio 4)

edad = int(input("Ingresa tu edad: "))

if edad <= 3:
    print("Bebé")
elif edad <= 12:
    print("Niño")
elif edad >= 13:
    print("Adolescente")
else:
    print("Adulto")
```

```python
#ejercicio 5
precio = 45 
tipo = "adulto_mayor"
if tipo == "adulto_mayor":
    descuento = precio*0.12
elif tipo == "profesor" or tipo == " estudiante":
    descuento = precio*0.10
else: 
    descuento = 0
total = precio-descuento
print(f"total a pagar ${total:.2f}")
```
