```python
#División de cuenta.

total_cuenta =float(input("Ingresa el total de la cuenta:"))
propina = float(input("Ingresa el porcentaje de propina que desear dejar:"))
personas = int(input("Ingresa la cantidad de personas en que se dividira la cuenta:"))

total_propina = total_cuenta * (propina/100)
total_cuenta_propina = total_cuenta + total_propina
total_persona = total_cuenta_propina / personas 

print(f"El total de la cuenta con propina es: ${total_cuenta_propina:.2f}")
print(f"La propina es:  ${total_propina:.2f}")
print(f"El total por persona es: ${total_persona:.2f}")
```

Para realizar la division de cuenta primero entendi lo que tenia que hacer que era sacra el porcentaje de propina y luego agregarselo para que al final pudiera dicidirlo entre la cantidad de personas, este codigo ya es funcionable para cualquier cuenta.

```python
#Conversor de minutos a días, horas y minutos.

minutos = int(input("Ingresa la cantidad de minutos que quieres convertir: "))

dias = minutos // 1440
horas =(minutos % 1440) //60
minutos_restantes = minutos % 60

print(f"{minutos} minutos equivalen a: {dias} día(s), {horas} hora(s) y {minutos_restantes} minuto(s).")
```

Para hacer este codigo para convertir minutos a horas, días y minutos restantes simplemnete utilice las pistas del profe, sabiendo que si los minutos los divides entre la cantidad de minutos que hay al día te arrojan un resultado y asi, como en las horas el residuo de la divion de minutos entre los minutos de el dia entre 60  daran las horas equivalente, al final utilice un f string ya que para mi es más facil acomodar así las variables que poner un print por cada una.

```python
#Calificación final ponderada

parcial1 = float(input("Primer parcial (30%):"))
parcial2 = float(input("Segundo parcial (30%):"))
parcial3 = float(input("Tercer parcial (40%):"))

calificacion_final = (parcial1 * 0.3) + (parcial2 * 0.3) + (parcial3 * 0.4)
print(f"La calificación final es: {calificacion_final:.2f}")
```

Para este ejercicio de sacar la calificacion ponderada simplemente lo pense como sacar un promedio pero haciendo unas ecuaciones antes ya que cada una de las calificaciones equivale a un porcentaje diferente y asifue como multiplique pir su equivalente y al final el promedio que salio.
